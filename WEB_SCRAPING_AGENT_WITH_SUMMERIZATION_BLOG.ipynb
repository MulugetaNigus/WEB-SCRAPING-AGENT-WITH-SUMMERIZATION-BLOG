```python
from crewai_tools import ScrapeWebsiteTool, Tool
from crewai import Agent, Task, Crew, Process
import os

# To enable scrapping any website it finds during it's execution
tool = ScrapeWebsiteTool()

# Initialize the tool with the website URL,
# so the agent can only scrap the content of the specified website

tool = ScrapeWebsiteTool(website_url='https://ebstv.tv')

# Extract the text from the site
text_tool = tool.run()
```

```python
summerizer = Agent(
    role = "summerize the content of the website",
    goal = "summerize the content that passed for you though tools array, always try to make your best to summerize the content",
    backstory="""You're a content summerizer at a large company.
    You're responsible for analyzing data and providing insights
    You're currently working on a project to summerize the content""",
    tools = [tool],
    verbose = True,
    allow_delegation = False
)

writer = Agent(
    role = "based on summerized content write a blog post.",
    goal = "based on summerized content write a blog post for the product with easy to follow",
    backstory="""You're a content writer and blogger at a large company.
    You're responsible for analyzing data and providing insights
    You're currently working on a project to summerize the content""",
    tools = [tool],
    verbose = True,
    allow_delegation=True
)
```

```python
task1 = Task(
    description='Find and summarize the latest and most relevant info provided by summerizer agent',
    agent=summerizer,
    expected_output='A bullet list summary of the scraped content'
)

task2 = Task(
    description='Find and summarize the latest and most relevant info provided by summerizer agent and write a blog post',
    agent=summerizer,
    expected_output='A bullet list summary of the scraped content and try to your best to write a blog post'
)

# save_output_task = Task(
#     description='Save the summarized AI news to a file',
#     expected_output='File saved successfully',
#     agent=writer,
#     tools=[file_save_tool],
#     output_file='outputs/Web_Summary.txt',
#     create_directory=True
# )
```

```python
crew = Crew(
    agents = [summerizer, writer],
    tasks = [task1, task2],
    verbose = True,
    process = Process.sequential
)

output = crew.kickoff()
print(output)
```
